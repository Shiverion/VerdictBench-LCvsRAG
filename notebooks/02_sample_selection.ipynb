# 02. Sample Selection

This notebook validates the stratified 50-verdict benchmark sample.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from notebooks.notebook_utils import *

set_plot_style()
ROOT = Path.cwd().resolve()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent

sample = load_sample()
meta = load_verdict_metadata()
merged = sample.merge(meta[['file_id', 'jenis_perkara', 'est_pages']], on='file_id', how='left', suffixes=('', '_meta'))
print(f'Sample size: {len(sample)}')
merged.head()
```

## Stratum balance

```python
sample['stratum'].value_counts().rename_axis('stratum').to_frame('count')
```

```python
sns.countplot(data=sample, x='stratum', order=['short', 'medium', 'long'], palette='crest')
plt.title('Benchmark sample by length stratum')
plt.tight_layout()
```

## Duplicate and existence checks

```python
pd.DataFrame({
    'duplicate_file_ids': [int(sample['file_id'].duplicated().sum())],
    'missing_metadata_rows': [int(merged['jenis_perkara'].isna().sum())],
})
```

## Case-type coverage

```python
merged['jenis_perkara'].fillna('unknown').value_counts().rename_axis('jenis_perkara').to_frame('count')
```

```python
sns.boxplot(data=sample, x='stratum', y='est_pages', order=['short', 'medium', 'long'], palette='Set2')
plt.title('Estimated pages by selected stratum')
plt.tight_layout()
```
