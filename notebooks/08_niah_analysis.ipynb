# 08. Needle-in-a-Haystack Analysis

This notebook evaluates deep-evidence questions and approximates needle depth by locating gold paragraphs inside the cleaned verdict text.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from notebooks.notebook_utils import *

set_plot_style()
ROOT = Path.cwd().resolve()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent

summary = pd.read_csv(ROOT / 'results/additional/niah/niah_summary.csv')
niah, niah_summary = load_niah_results()
depth_summary = niah_by_depth()
summary
```

## NIAH condition summary

```python
markdown_df(niah_summary[['architecture', 'queries', 'mean_faithfulness', 'mean_bertscore_f1', 'zero_faithfulness_cases']])
```

```python
summary_plot = summary.assign(condition=summary['condition'].replace({'simple_rag_cs512_k5_fixed':'Simple RAG', 'advanced_rag_QR_MF_HS_RR':'Advanced RAG', 'lc':'Long Context'}))
sns.barplot(data=summary_plot, x='condition', y='niah_accuracy', palette='crest')
plt.title('NIAH accuracy by architecture')
plt.tight_layout()
```

## Approximate depth buckets

```python
markdown_df(depth_summary)
```

```python
depth_plot = depth_summary[depth_summary['depth_bucket'] != 'unknown']
if not depth_plot.empty:
    sns.barplot(data=depth_plot, x='depth_bucket', y='mean_faithfulness', hue='architecture', order=['shallow', 'deep', 'bottom'])
    plt.title('NIAH faithfulness by approximate needle depth')
    plt.tight_layout()
else:
    print('Depth buckets could not be reconstructed from the available cleaned texts.')
```
