# 06. Ablation Analysis

Component-level analysis of the Advanced RAG pipeline.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from notebooks.notebook_utils import *

set_plot_style()
ROOT = Path.cwd().resolve()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent

ablation, ablation_summary = load_ablation_results()
markdown_df(ablation_summary[['condition_label', 'queries', 'mean_faithfulness', 'mean_bertscore_f1', 'mean_hallucination_rate', 'mean_context_precision', 'mean_total_cost_usd']])
```

## Faithfulness deltas versus baseline

```python
baseline = float(ablation_summary.loc[ablation_summary['condition_label'] == 'Baseline Simple RAG', 'mean_faithfulness'].iloc[0])
deltas = ablation_summary[['condition_label', 'mean_faithfulness']].copy()
deltas['delta_vs_baseline'] = deltas['mean_faithfulness'] - baseline
markdown_df(deltas)
```

```python
sns.barplot(data=deltas, y='condition_label', x='delta_vs_baseline', palette='coolwarm')
plt.axvline(0, color='black', linewidth=1)
plt.title('Faithfulness delta vs Simple RAG baseline')
plt.tight_layout()
```

## Incremental metric view

```python
metric_cols = ['mean_faithfulness', 'mean_hallucination_rate', 'mean_context_precision']
ablation_summary.set_index('condition_label')[metric_cols].round(3)
```

```python
sns.lineplot(data=ablation_summary, x='condition_label', y='mean_faithfulness', marker='o')
plt.xticks(rotation=30, ha='right')
plt.title('Faithfulness across ablation stages')
plt.tight_layout()
```
